# Executive Summary

This analysis examined 7,043 telecom customers to identify the primary drivers of customer churn and their financial impact.

## Key Findings

• Overall churn rate is XX%.

• Month-to-month contracts experience the highest churn.

• Customers with short tenure are significantly more likely to leave.

• Customers without Online Security and Premium Tech Support exhibit higher churn.

• Fiber Optic customers have higher churn than other internet types.

• A small percentage of cities contribute disproportionately to churn.

• Churn results in substantial revenue loss.

• Long-tenure customers generate considerably higher lifetime revenue.

# Recommendations

### 1. Improve Early Customer Retention

Focus on customers within their first 12 months.

---

### 2. Promote Long-Term Contracts

Offer discounts for one-year and two-year contracts.

---

### 3. Bundle Online Security

Customers with Online Security churn less.

Include it in promotional packages.

---

### 4. Expand Premium Tech Support

Increase adoption through free trial periods.

---

### 5. Target High-Risk Cities

Launch localized retention campaigns.

---

### 6. Protect High-Value Customers

Identify customers with high revenue before they churn.

Offer loyalty rewards.

---

### 7. Encourage Auto Payments
